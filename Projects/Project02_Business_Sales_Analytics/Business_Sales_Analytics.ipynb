````python
import pandas as pd
sales = pd.read_csv(r"C:\Users\manid\OneDrive\Desktop\Project-Atlas\Datasets\Superstore_sales\SampleSuperstore.csv")
#Inspecting First and last record
print("First Record : \n",sales.head(1))
print("\nLast Record : \n",sales.tail(1))
#Dimensions and Datatypes
print("\nDimensions : ",sales.ndim)
print("\nDataTypes : \n",sales.dtypes)
````

# 🔍 Part 3 – Exploratory Data Analysis (EDA)

## Objectives

Explore the overall characteristics of the dataset by answering questions such as:

- Total Sales
- Total Profit
- Total Quantity Sold
- Average Sales
- Average Profit
- Average Discount
- Number of Cities

````python
#Total Sales
print("Total Sales : ",sales["Sales"].sum())
#Total Profit
print("\nTotal Profit : ",sales["Profit"].sum())
#Total Quantity Sold
print("\nTotal Quantity Sold : ",sales["Quantity"].sum())
#Average Sales
print("\nAverage Sales : ",sales["Sales"].mean())
#Average Profit
print("\nAverage Profit : ",sales["Profit"].mean())
#Average Discount
print("\nAverage Discount : ",sales["Discount"].mean())
#Number of Cities
print("\nNumber of Cities : ",sales["City"].nunique())
#Number of States
print("\nNumber of States : ",sales["State"].nunique())
#Number of Categories
print("\nNumber of Categories : ",sales["Category"].nunique())
#Number of Sub-Categories
print("\nNumber of Sub-Categories : ",sales["Sub-Category"].nunique())
````

# 📊 Part 4 – Category & Segment Analysis

## Objectives

Analyze business performance across:

- Categories
- Sub-Categories
- Customer Segments

Determine:

- Highest Sales
- Highest Profit
- Lowest Profit
- Most Sold Category
- Most Profitable Segment

Deliverable:
Identify which business areas contribute most to revenue and profitability.

````python
#Highest Sales
print("Highest Sales by Categories : ",sales.groupby(by="Category")["Sales"].sum().idxmax())
print("\nHighest Sales by Sub-Categories : ",sales.groupby(by="Sub-Category")["Sales"].sum().idxmax())
print("\nHighest Sales by Customer Segments : ",sales.groupby(by="Segment")["Sales"].sum().idxmax())
#Highest Profit
print("Highest Profit by Categories : ",sales.groupby(by="Category")["Profit"].sum().idxmax())
print("\nHighest Profit by Sub-Categories : ",sales.groupby(by="Sub-Category")["Profit"].sum().idxmax())
print("\nHighest Profit by Customer Segments : ",sales.groupby(by="Segment")["Profit"].sum().idxmax())
#Lowest Profit
print("Lowest Profit by Categories : ",sales.groupby(by="Category")["Profit"].sum().idxmin())
print("\nLowest Profit by Sub-Categories : ",sales.groupby(by="Sub-Category")["Profit"].sum().idxmin())
print("\nLowest Profit by Customer Segments : ",sales.groupby(by="Segment")["Profit"].sum().idxmin())
#Most Sold Category
print("Most Sold Category : ",sales.groupby(by="Category")["Quantity"].max().idxmax())
#Most Profitable Segment
print("\nMost Profitable Segment : ",sales.groupby(by="Segment")["Profit"].max().idxmax())
````

# 🚚 Part 6 – Shipping Analysis

## Objectives

Analyze shipping performance.

Determine:

- Most frequently used Ship Mode
- Average Sales by Ship Mode
- Average Profit by Ship Mode
- Total Orders by Ship Mode

Deliverable:
Understand shipping preferences and profitability.

````python
#Most frequently used Ship Mode
print("Most frequently used Ship Mode : ",sales.groupby(by="Ship Mode")["Segment"].count().idxmax())
#Average Sales by Ship Mode
print("\nAverage Sales by Ship Mode : \n",sales.groupby(by="Ship Mode")["Sales"].mean())
#Average Profit by Ship Mode
print("\nAverage Profit by Ship Mode : \n",sales.groupby(by="Ship Mode")["Profit"].mean())
#Total Orders by Ship Mode
print("\nTotal Orders by Ship Mode : \n",sales.groupby(by="Ship Mode")["Segment"].count())
````

# 💰 Part 7 – Discount Analysis

## Objectives

Investigate how discounts impact profitability.

Determine:

- Average Discount
- Highest Discount
- Average Profit by Discount
- Whether larger discounts reduce profitability

Deliverable:
Evaluate the effectiveness of discount strategies.

````python
#Average Discount
print("Average Discount : ",sales["Discount"].mean())
#Highest Discount
print("\nHighest Discount : ",sales["Discount"].max())
#Average Profit by Discount
print("\nAverage Profit by Discount : \n",sales.groupby(by="Discount")["Profit"].mean())
#Whether larger discounts reduce profitability
print("\nWhether larger discounts reduce profitability : ",sales["Discount"].corr(sales["Profit"]))
````

## The Statistical Meaning
- It is Negative (-): The minus sign confirms an inverse relationship. As the discount percentage goes up, the profit goes down.
- It is Weak to Moderate (0.22): Because the number is closer to 0 than to -1, the relationship isn't perfectly strictly linear. This means while discounts definitely hurt profits, they aren't the only factor dragging profit down (things like high shipping costs, low initial margins, or specific product categories are also playing a major role).
