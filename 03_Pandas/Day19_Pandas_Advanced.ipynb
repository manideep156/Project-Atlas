# 📂 Practice Dataset

Today's dataset represents monthly sales performance of employees across different regions. It will be used to practice advanced sorting, ranking, filtering, and efficient data manipulation techniques.

````python
import pandas as pd

sales = pd.DataFrame({

    "Employee":[
        "Alice","Bob","Charlie","David","Eva",
        "Frank","Grace","Helen","Ian","Jack"
    ],

    "Region":[
        "East","West","North","South","East",
        "West","North","South","East","West"
    ],

    "Department":[
        "Electronics","Furniture","Electronics","Accessories","Furniture",
        "Electronics","Accessories","Furniture","Electronics","Accessories"
    ],

    "Sales":[
        85000,72000,96000,65000,78000,
        91000,69000,81000,99000,73000
    ],

    "Orders":[
        120,98,140,90,110,
        135,95,115,150,100
    ]

})

print(sales)
````

# 📖 Part 1 – Sorting Data

**Objective:** Organize datasets efficiently using sorting operations.

## ✅ Tasks

Using the Practice Dataset:

- Sort by Sales (Ascending).
- Sort by Sales (Descending).
- Sort by Region and Sales.
- Sort by Orders.
- Compare different sorting strategies.

````python
#Sort by Sales (Ascending).
print("Sort by Sales (Ascending) : \n",sales.sort_values(by="Sales"))
#Sort by Sales (Descending).
print("\nSort by Sales (Descending) : \n",sales.sort_values(by="Sales",ascending=False))
````

````python
#Sort by Region and Sales.
print("\nSort by Region and Sales : \n",sales.sort_values(by=["Region","Sales"]))
#Sort by Orders.
print("\nSort by Orders : \n",sales.sort_values(by="Orders"))
````

### 📝 Observation

- Sorting organizes records for easier analysis.
- Multiple-column sorting improves data interpretation.
- Sorting is commonly used before reporting and visualization.

# 📖 Part 2 – Ranking Data

**Objective:** Assign rankings based on numerical values.

## ✅ Tasks

- Rank employees by Sales.
- Rank employees by Orders.
- Compare dense ranking and standard ranking.
- Find the Top 3 performers.
- Identify the lowest-ranked employee.

````python
sales['Sales_Rank'] = sales['Sales'].rank(ascending=False).astype(int)
sales['Orders_Rank'] = sales['Orders'].rank(ascending=False).astype(int)
# 'min' is standard competition ranking (1, 2, 2, 4). 'dense' leaves no gaps (1, 2, 2, 3).
sales['Standard_Sales_Rank'] = sales['Sales'].rank(method='min', ascending=False).astype(int)
sales['Dense_Sales_Rank'] = sales['Sales'].rank(method='dense', ascending=False).astype(int)
#Find the Top 3 performers (Based on Sales)
top_3_performers = sales.nlargest(3, 'Sales')

#Identify the lowest-ranked employee (Based on Sales)
lowest_ranked_employee = sales.nsmallest(1, 'Sales')

print("--- Full Dataset with Rankings ---")
print(sales[['Employee', 'Sales', 'Sales_Rank', 'Orders_Rank', 'Standard_Sales_Rank', 'Dense_Sales_Rank']].sort_values('Sales_Rank'))

print("\n--- Top 3 Performers (Sales) ---")
print(top_3_performers[['Employee', 'Sales', 'Sales_Rank']])

print("\n--- Lowest-Ranked Employee (Sales) ---")
print(lowest_ranked_employee[['Employee', 'Sales', 'Sales_Rank']])
````

### 📝 Observation

- Ranking highlights business performance.
- Different ranking methods handle ties differently.
- Rankings simplify leaderboards and KPI reporting.

# 📖 Part 3 – Selecting Top Records

**Objective:** Retrieve the highest and lowest values efficiently.

## ✅ Tasks

- Find the top 5 sales.
- Find the bottom 3 sales.
- Highest order count.
- Lowest order count.
- Compare with sorting.

````python
#Find the top 5 sales.
print("Find the top 5 sales : \n",sales.nlargest(5,"Sales"))
#Find the bottom 3 sales.
print("\nFind the bottom 3 sales : \n",sales.nsmallest(3,"Sales"))
````

````python
#Highest order count.
print("Highest order count : \n",sales.nlargest(1,"Orders"))
#Lowest order count.
print("\nLowest order count : \n",sales.nsmallest(1,"Orders"))
````

### 📝 Observation

- `nlargest()` and `nsmallest()` are faster than full sorting.
- These functions are useful for dashboards and business reports.
- Selecting top records is a common analytical requirement.

# 📖 Part 4 – Querying Data

**Objective:** Filter data using expressive query syntax.

## ✅ Tasks

- Employees with Sales > 80000.
- Electronics department only.
- East region with Sales > 90000.
- Furniture department with Orders > 100.
- Multiple-condition filtering using `query()`.

````python
#Employees with Sales > 80000.
print("Employees with Sales > 80000 : \n",sales.query("Sales>80000").Employee)
#Electronics department only.
print("\nElectronics department only : \n",sales.query("Department == 'Electronics'").Employee)
#East region with Sales > 90000.
print("\nEast region with Sales > 90000 : \n",sales.query("Region == 'East' & Sales>90000").Employee)
#Furniture department with Orders > 100.
print("\nFurniture department with Orders > 100 : \n",sales.query("Department == 'Furniture' and Orders>100").Employee)
````

### 📝 Observation

- `query()` improves readability for complex filters.
- It reduces lengthy boolean expressions.
- Querying simplifies business rule implementation.

# 📖 Part 5 – Evaluating Expressions

**Objective:** Create calculated columns efficiently.

## ✅ Tasks

Create:

- Average Order Value.
- Performance Score.
- Bonus Eligibility.
- High Performer Flag.
- Compare `eval()` with standard column operations.

````python
#Average Order Value.
sales = sales.eval("Average_ODV = Sales/Orders")
#Performance Score.
sales = sales.eval("Performance_Score = Orders/Orders.max()")
#Bonus Eligibility.
sales = sales.eval("Bonus_Eligibility = Performance_Score>0.7")
#High Performer Flag.
sales = sales.eval("High_Performer = Performance_Score>0.85")
print(sales)
````

### 📝 Observation

- `eval()` improves readability for mathematical expressions.
- Calculated columns support business reporting and KPI generation.
- Efficient expressions become more valuable as datasets grow.

# 🧩 Mini Challenge – Employee Performance Dashboard

## Scenario

Create a dataset containing:

- Employee
- Department
- Monthly Sales
- Customer Rating
- Orders Completed

Using Pandas:

- Rank employees by Monthly Sales.
- Find the Top 5 performers.
- Retrieve the Bottom 2 performers.
- Filter employees with ratings above 4.5.
- Create a Performance Score column.
- Identify bonus-eligible employees.

Finally,

Write a short summary explaining the overall employee performance.

````python
# Create the dataset using a dictionary
data = {
    "Employee": [
        "Sarah Jenkins", "Marcus Thorne", "Priya Patel", "James O'Connor", 
        "Elena Rodriguez", "David Kim", "Anita Desai", "Michael Chang", 
        "Rachel Foster", "Thomas Wright"
    ],
    "Department": [
        "Electronics", "Appliances", "Home Goods", "Electronics", 
        "Appliances", "Home Goods", "Outdoor", "Electronics", 
        "Outdoor", "Appliances"
    ],
    "Monthly_Sales": [
        42500, 38100, 15400, 51200, 29800, 
        18900, 31500, 48000, 27300, 41900
    ],
    "Customer_Rating": [
        4.8, 4.5, 4.9, 4.2, 4.7, 
        4.4, 4.9, 4.1, 4.6, 4.8
    ],
    "Orders_Completed": [
        145, 92, 210, 156, 78, 
        185, 115, 134, 98, 105
    ]
}

# Load the data into a pandas DataFrame
df = pd.DataFrame(data)

# Display the dataset
print(df)
````

````python
df["Monthly Sales Ranking"] = df["Monthly_Sales"].rank().astype(int)
print(df)
#Top 5 performers
print("\nTop 5 performers : \n",df.nlargest(5,["Customer_Rating","Orders_Completed"]))
````

````python
#Bottom 2 Performers
print("\nBottom 2 Performers : \n",df.nsmallest(2,["Customer_Rating","Orders_Completed"]))
#Employees with rating > 4.5
print("\nEmployees with rating>4.5 : \n",df.query("Customer_Rating > 4.5").Employee)
````

````python
#Performance score
df=df.eval("Performance_Score = Orders_Completed * Customer_Rating /1000")
df=df.eval("Bonus_Eligible = Performance_Score > Performance_Score.mean()")
print(df)
````

## 📌 Mini Challenge Summary

Using an employee performance dataset, I ranked employees based on monthly sales, identified the top and bottom performers, filtered employees with customer ratings above 4.5, created a Performance Score metric, and determined bonus eligibility based on business criteria.

The analysis highlighted high-performing employees, recognized consistent customer satisfaction, and demonstrated how ranking, filtering, and calculated columns can support performance evaluations and incentive programs in an organization.

# 🚀 Bonus Challenge – E-Commerce Product Performance

## Scenario

Create a dataset containing:

- Product
- Category
- Revenue
- Units Sold
- Customer Rating

Using Pandas:

- Rank products by Revenue.
- Find the Top 3 products.
- Retrieve the Bottom 3 products.
- Filter products with ratings above 4.5.
- Create Revenue per Unit.
- Identify premium-performing products.

Finally,

Write five analytical insights from your analysis.

````python

# Define the dataset as a dictionary
data = {
    "Product": [
        "Wireless Noise-Canceling Earbuds", "Ergonomic Mesh Office Chair", 
        "Stainless Steel Water Bottle", "Smart Fitness Watch", 
        "Non-Slip Yoga Mat", "Mechanical Gaming Keyboard", 
        "Organic Cotton T-Shirt", "Ceramic Coffee Mug Set", 
        "Adjustable Standing Desk", "High-Speed Blender"
    ],
    "Category": [
        "Electronics", "Furniture", "Home & Kitchen", "Electronics", 
        "Fitness", "Electronics", "Apparel", "Home & Kitchen", 
        "Furniture", "Appliances"
    ],
    "Revenue": [45000, 37500, 8400, 62000, 5600, 18200, 12500, 4200, 85000, 22400],
    "Units Sold": [300, 150, 420, 248, 160, 140, 500, 120, 170, 112],
    "Customer Rating": [4.6, 4.2, 4.8, 4.5, 4.7, 4.4, 4.3, 4.9, 4.5, 4.1]
}

# Create the DataFrame
df = pd.DataFrame(data)
print(df)
````

````python
#Rank products by Revenue.
df["Revenue_Ranking"] = df["Revenue"].rank().astype(int)
print(df)
#Find the Top 3 products.
print("\nFind the Top 3 products : \n",df.nlargest(3,["Revenue","Units Sold","Customer Rating"]).Product)
````

````python
#Retrieve the Bottom 3 products.
print("\nFind the Bottom 3 products : \n",df.nsmallest(3,["Revenue","Units Sold","Customer Rating"]).Product)
#Filter products with ratings above 4.5.
print("\nproducts with ratings above 4.5 : \n",df.query("`Customer Rating`>4.5"))
````

````python
#Create Revenue per Unit.
df=df.eval("Revenue_per_unit = Revenue/`Units Sold`")
print(df)
````

## 💡 Analytical Insights

1. **The Adjustable Standing Desk generated the highest revenue ($85,000)**, making it the top-performing product despite selling only 170 units. This indicates a high-value product with strong revenue potential.

2. **The Organic Cotton T-Shirt sold the highest number of units (500)** but generated only $12,500 in revenue, suggesting it is a high-volume, low-margin product.

3. **The Ceramic Coffee Mug Set received the highest customer rating (4.9)** even though it had the lowest revenue ($4,200), indicating strong customer satisfaction but relatively low market demand.

4. **Electronics products consistently contributed high revenue**, with the Smart Fitness Watch ($62,000) and Wireless Noise-Canceling Earbuds ($45,000) performing particularly well, making Electronics the strongest revenue-generating category.

5. **Revenue per unit varied significantly across products**, showing that premium products such as the Adjustable Standing Desk generated substantially more revenue per sale than high-volume products like the Organic Cotton T-Shirt. This insight can help guide pricing and product strategy.

# 🌟 Analyst Extension Challenge – Answers

### 1. Sort employees by sales in descending order.

**Answer:**

```python
df.sort_values(by="Sales", ascending=False)
```

**Reason:**
Sorts the DataFrame from the highest to the lowest sales value.

---

### 2. Retrieve only the five highest revenue records.

**Answer:**

```python
df.nlargest(5, "Revenue")
```

**Reason:**
Efficiently returns the top five records without sorting the entire dataset.

---

### 3. Assign ranks to products based on revenue.

**Answer:**

```python
df["Revenue_Rank"] = df["Revenue"].rank(ascending=False)
```

**Reason:**
Assigns a numerical rank to each product according to its revenue.

---

### 4. Filter records where Sales > 80,000 and Region is East.

**Answer:**

```python
df.query("Sales > 80000 and Region == 'East'")
```

**Reason:**
Uses the `query()` method to perform readable multi-condition filtering.

---

### 5. Create a new calculated column using an arithmetic expression.

**Answer:**

```python

```

**Reason:**
Creates a new calculated column efficiently using `eval()`.

---

### 6. Retrieve the three smallest order values.

**Answer:**

```python
df.nsmallest(3, "Orders")
```

**Reason:**
Returns the three records with the lowest order values efficiently.

---

### 7. Compare two ranking methods when duplicate values exist.

**Answer:**

```python
df["Standard_Rank"] = df["Revenue"].rank(method="average")

df["Dense_Rank"] = df["Revenue"].rank(method="dense")
```

**Reason:**

- **Standard Rank (`average`)** assigns the average rank to tied values and leaves gaps in subsequent rankings.
- **Dense Rank (`dense`)** assigns the same rank to tied values but does not leave gaps.

---

### 8. Determine the fastest approach to retrieve only the top records without sorting the entire dataset.

**Answer:**

```python
df.nlargest(5, "Revenue")
```

**Reason:**
`nlargest()` is optimized for retrieving only the highest records and is generally more efficient than sorting the entire DataFrame with `sort_values()`.
