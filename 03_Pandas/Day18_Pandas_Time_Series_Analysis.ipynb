# 📂 Practice Dataset

Today's dataset contains daily retail sales for one month. The Date column will be converted into the DataFrame index to enable efficient time-series operations.

````python
import pandas as pd

sales = pd.DataFrame({

    "Date":pd.date_range("2026-01-01", periods=30),

    "Revenue":[
        42000,43500,44100,43800,45000,
        46300,47000,48200,47800,48900,
        50100,49600,51000,52300,51900,
        53100,54000,54800,55200,56100,
        57000,56500,57800,59000,59600,
        60200,61100,62000,62800,63500
    ]

})

print(sales)
````

# 📖 Part 1 – Setting the Datetime Index

**Objective:** Learn why datetime indexes are important for time-series analysis.

## ✅ Tasks

- Convert the Date column into the DataFrame index.
- Verify the index type.
- Observe how the DataFrame changes.
- Explain the advantages of a datetime index.
- Compare integer indexing and datetime indexing.

````python
sales = sales.set_index("Date")
print(sales)
print("Index Datatype",sales.index.dtype)
````

### 📝 Observation

- Datetime indexes enable efficient chronological operations.
- Time-based indexing becomes more intuitive than integer indexing.
- Most time-series functions in Pandas require a datetime index.

# 📖 Part 2 – Time-Based Indexing

**Objective:** Access data using dates instead of row numbers.

## ✅ Tasks

Retrieve:

- One specific day.
- A range of dates.
- The first week.
- Transactions after January 20.
- The final three days.

````python
specific_day = sales.loc["2026-01-15"]
print(specific_day, "\n")


date_range = sales.loc["2026-01-10":"2026-01-15"]
print(date_range, "\n")


first_week = sales.loc[:"2026-01-07"]
print(first_week, "\n")
````

````python
after_jan_20 = sales.loc["2026-01-21":]
print(after_jan_20, "\n") 


final_three_days = sales.tail(3)
print(final_three_days, "\n")
````

### 📝 Observation

- Datetime indexes support intuitive slicing using dates.
- Date-based indexing improves readability and maintainability.
- Analysts frequently use time-based indexing for reporting.

# 📖 Part 3 – Resampling Data

**Objective:** Summarize data across different time intervals.

## ✅ Tasks

Calculate:

- Weekly revenue.
- Weekly average revenue.
- Monthly revenue.
- Monthly average revenue.
- Compare daily and weekly summaries.

````python
# Calculate Weekly Revenue (Total sum per week)
# 'W' stands for Weekly (defaults to ending on Sunday)
weekly_revenue = sales.resample('W').sum()
print("--- Weekly Revenue ---")
print(weekly_revenue, "\n")

# Calculate Weekly Average Revenue (Mean per week)
weekly_avg_revenue = sales.resample('W').mean()
print("--- Weekly Average Revenue ---")
print(weekly_avg_revenue, "\n")

# Calculate Monthly Revenue (Total sum per month)
# 'ME' stands for Month-End (in older pandas versions, use 'M')
monthly_revenue = sales.resample('ME').sum()
print("--- Monthly Revenue ---")
print(monthly_revenue, "\n")

# Calculate Monthly Average Revenue (Mean per month)
monthly_avg_revenue = sales.resample('ME').mean()
print("--- Monthly Average Revenue ---")
print(monthly_avg_revenue, "\n")
````

### 📝 Observation

- Resampling groups observations into meaningful time intervals.
- Weekly and monthly summaries simplify trend analysis.
- Resampling is widely used in financial and business reporting.

# 📖 Part 4 – Rolling Statistics

**Objective:** Smooth fluctuations using rolling calculations.

## ✅ Tasks

Calculate:

- 3-day moving average.
- 5-day moving average.
- Rolling maximum.
- Rolling minimum.
- Compare rolling averages with original revenue.

````python
sales["3-Day MA"] = sales["Revenue"].rolling(window=3).mean()
sales["5-Day MA"] = sales["Revenue"].rolling(window=5).mean()
sales["5-Day Max"] = sales["Revenue"].rolling(window=5).max()
sales["5-Day Min"] = sales["Revenue"].rolling(window=5).min()


print("--- 1. Original Revenue vs 3-Day Moving Average ---")
# Notice how the first 2 days are NaN, and it starts calculating on Day 3
print(sales[["Revenue", "3-Day MA"]].head(7), "\n")


print("--- 2. Original Revenue vs 5-Day Moving Average ---")
# Notice how the first 4 days are NaN, and it starts calculating on Day 5
print(sales[["Revenue", "5-Day MA"]].head(7), "\n")


print("--- 3. Rolling 5-Day Maximum ---")
print(sales[["Revenue", "5-Day Max"]].head(7), "\n")


````

````python
print("--- 4. Rolling 5-Day Minimum ---")
print(sales[["Revenue", "5-Day Min"]].head(7), "\n")


print("--- 5. The Complete DataFrame (First 10 Days) ---")
print(sales.head(10))
````

### 📝 Observation

- Rolling statistics reduce short-term fluctuations.
- Moving averages reveal long-term business trends.
- Rolling windows are widely used in forecasting and anomaly detection.

# 📖 Part 5 – Trend Analysis

**Objective:** Analyze business performance over time.

## ✅ Tasks

Identify:

- Highest revenue day.
- Lowest revenue day.
- Overall growth.
- Daily revenue change.
- Days where revenue increased compared to the previous day.

````python
print("Highest Revenue Day : ",sales["Revenue"].idxmax())
print("\nLowest Revenue Day : ",sales["Revenue"].idxmin())
sales["Daily Change"] = sales["Revenue"].diff()
print(sales[["Revenue", "Daily Change"]].head(), "\n")
````

````python
print("Days where revenue increased compared to the prev day : \n",sales[sales["Daily Change"]>0])
````

### 📝 Observation

- Trend analysis measures business growth over time.
- Comparing consecutive days helps identify performance changes.
- Trend metrics support forecasting and strategic planning.

# 🧩 Mini Challenge – Website Traffic Analysis

## Scenario

A company's analytics team monitors daily website visitors.

Create a dataset containing:

- Date
- Visitors
- Bounce Rate
- Average Session Duration

Using Pandas:

- Convert Date into a datetime index.
- Calculate weekly average visitors.
- Find the busiest day.
- Compute a 7-day moving average for visitors.
- Identify days where visitor count increased compared to the previous day.

Finally,

Write a short summary describing website traffic trends.

````python
data = {
    "Date": [
        "2026-07-20", "2026-07-21", "2026-07-22", 
        "2026-07-23", "2026-07-24", "2026-07-25", "2026-07-26"
    ],
    "Visitors": [2450, 2610, 2390, 2805, 3100, 1850, 1920],
    "Bounce_Rate_Percent": [45.2, 43.8, 46.1, 42.5, 41.0, 52.4, 51.1],
    "Avg_Session_Duration": [
        "02:15", "02:30", "02:05", "02:45", "03:10", "01:50", "01:55"
    ]
}

df = pd.DataFrame(data)

df['Date'] = pd.to_datetime(df['Date'])
df=df.set_index("Date")

print(df)
````

````python
print("Weekly average visitors : ",df["Visitors"].resample('W').mean())
print("\nBusiest Day : ",df["Visitors"].idxmax())
print("\n7 day moving average : ",df["Visitors"].rolling(7).mean())
df["Daily Change"] = df["Visitors"].diff()
print("\n days where visitors are greater than prev day : \n",df[df["Daily Change"]>0].index)
````

## 📌 Mini Challenge Summary

Using a website traffic dataset, I converted the Date column into a datetime index and analyzed visitor trends over a one-week period. I calculated weekly average visitors, identified the busiest day, computed a 7-day moving average, and tracked daily visitor growth.

This analysis demonstrated how time-series techniques help monitor website performance, identify traffic patterns, and support data-driven decisions for improving user engagement.

# 🚀 Bonus Challenge – Electricity Consumption Analysis

## Scenario

Create a dataset containing:

- Date
- Electricity Usage (kWh)
- Peak Demand
- Cost

Using Pandas:

- Convert Date into a datetime index.
- Calculate weekly electricity consumption.
- Compute a 5-day moving average.
- Identify the highest consumption day.
- Calculate daily usage changes.
- Compare weekly consumption trends.

Finally,

Write five analytical insights from your analysis.

````python
data = {
    'Date': [
        '2026-07-01', '2026-07-02', '2026-07-03', '2026-07-04', '2026-07-05',
        '2026-07-06', '2026-07-07', '2026-07-08', '2026-07-09', '2026-07-10'
    ],
    'Electricity Usage (kWh)': [
        45.2, 42.8, 51.5, 38.9, 36.4, 
        47.1, 49.3, 53.0, 44.6, 41.2
    ],
    'Peak Demand (kW)': [
        5.1, 4.8, 6.2, 4.1, 3.9, 
        5.5, 5.8, 6.5, 5.0, 4.6
    ],
    'Cost': [
        6.78, 6.42, 7.73, 5.84, 5.46, 
        7.07, 7.40, 7.95, 6.69, 6.18
    ]
}

df = pd.DataFrame(data)

df['Date'] = pd.to_datetime(df['Date'])
df = df.set_index("Date")
print(df)
````

````python
print("Weekly electrical consumption : \n",df["Electricity Usage (kWh)"].resample('W').sum())
print("\n5 day moving average : \n",df["Electricity Usage (kWh)"].rolling(5).mean())
print("\nHighest Electricity Consumption : ",df["Electricity Usage (kWh)"].idxmax())
print("\nDaily Usage Changes : ",df["Electricity Usage (kWh)"].diff())
````

## 💡 Analytical Insights

1. **July 8 recorded the highest electricity usage (53.0 kWh)** along with the highest peak demand (6.5 kW), indicating the period of maximum energy consumption.

2. **Electricity usage and total cost showed a strong positive relationship**, meaning higher energy consumption directly resulted in increased operational costs.

3. **The 5-day moving average smoothed daily fluctuations**, making it easier to observe the overall consumption trend instead of focusing on individual daily variations.

4. **Weekly consumption analysis highlighted periods of consistently higher energy usage**, helping identify potential peak operating days for better energy planning.

5. **Daily usage differences revealed both increases and decreases in consumption**, providing valuable insights for detecting unusual spikes and monitoring energy efficiency over time.
