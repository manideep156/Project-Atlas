# 📖 Part 1 – Understanding Datetime Objects

**Objective:** Understand why date columns should be stored as datetime objects instead of plain text.

## ✅ Tasks

Using the Practice Dataset:

- Inspect the data type of the Date column.
- Explain why it is treated as text.
- Identify problems caused by storing dates as strings.
- Predict which analytical operations require datetime objects.
- Compare string dates with true datetime values.

````python
print(sales["Date"].dtype)
````

* **Why Text?** The data was provided as standard Python string literals, and pandas defaults to literal interpretation unless explicitly told to parse dates.
* **Problems with Strings:** You cannot perform date math (like finding days between sales), extracting parts like the month requires clunky string slicing, and sorting relies on perfect text formatting rather than actual chronological order.
* **Operations Requiring Datetime:** Time-series aggregation (like `.resample()` for monthly revenue), rolling averages, extracting calendar features (like `.dt.day_name()`), and time-based filtering.
* **String vs. Datetime:**
* **String:** Text format, requires manual string manipulation, fails on date math.
* **Datetime:** 64-bit integer format, unlocks the `.dt` accessor for instant calendar components, and fully supports time math and logical sorting.

### 📝 Observation

- The Date column is initially stored as a string (`object`) rather than a datetime object.
- String-based dates limit chronological analysis, filtering, sorting, and date calculations.
- Converting dates into datetime format enables powerful time-based operations in Pandas.

# 📖 Part 2 – Converting to Datetime

**Objective:** Convert text-based dates into Pandas datetime objects.

## ✅ Tasks

Using the Practice Dataset:

- Convert the Date column using `pd.to_datetime()`.
- Verify the updated data type.
- Compare the dataset before and after conversion.
- Observe how Pandas stores datetime values internally.
- Explain the benefits of datetime conversion.

````python
sales["Date"]=pd.to_datetime(sales["Date"])
print(sales["Date"].dtype)
````

### 📝 Observation

- `pd.to_datetime()` converts text into datetime objects without changing the actual date values.
- Datetime columns support chronological operations that are not possible with plain strings.
- Converting dates early improves data quality and simplifies future analysis.

# 📖 Part 3 – Extracting Date Components

**Objective:** Learn how to extract useful information from datetime columns.

## ✅ Tasks

After converting the Date column:

Extract:

- Year
- Month
- Day
- Month Name
- Day Name
- Quarter

Create separate columns for each extracted component and observe how these features simplify business reporting.

````python
sales["Year"] = sales["Date"].dt.year
sales["Month"] = sales["Date"].dt.month
sales["Day"] = sales["Date"].dt.day
sales["Month Name"] = sales["Date"].dt.month_name()
sales["Day Name"] = sales["Date"].dt.day_name()
sales["Quarter"] = sales["Date"].dt.quarter
print(sales)
````

### 📝 Observation

- The `.dt` accessor provides quick access to individual date components.
- Extracted features make grouping, filtering, and reporting much easier.
- Date components are commonly used in dashboards, trend analysis, and forecasting.

# 📖 Part 4 – Filtering with Dates

**Objective:** Retrieve records using date-based conditions.

## ✅ Tasks

Using the Practice Dataset:

Find:

- Sales after February 1, 2026.
- Sales during March.
- Sales before April.
- Quarter 1 transactions.
- Revenue generated during January.

````python
# Sales after February 1, 2026.
print("Sales after February 1, 2026 : \n", sales[sales["Date"] > "2026-02-01"])

# Sales during March.
print("\nSales during March : \n", sales[sales["Month"] == 3])
````

````python
# Sales before April.
print("\nSales before April : \n", sales[sales["Date"] < "2026-04-01"])

# Quarter 1 transactions.
# You can use the built-in .dt.quarter property
print("\nQuarter 1 transactions : \n", sales[sales["Date"].dt.quarter == 1])

# Revenue generated during January.
jan_revenue = sales.loc[sales["Month"] == 1, "Revenue"].sum()
print("\nRevenue generated during January : ", jan_revenue)
````

### 📝 Observation

- Datetime columns allow accurate filtering using comparison operators.
- Date filtering helps answer business questions involving specific periods.
- Time-based filtering is essential for reporting, auditing, and trend analysis.

# 📖 Part 5 – Date Arithmetic

**Objective:** Perform calculations using datetime values.

## ✅ Tasks

Using the Practice Dataset:

- Find the earliest transaction date.
- Find the latest transaction date.
- Calculate the number of days between each transaction and the first transaction.
- Create a new column showing elapsed days.
- Identify the oldest and most recent sales records.

````python
# 1. Find the earliest and latest transaction dates
earliest_date = sales["Date"].min()
latest_date = sales["Date"].max()

print(f"Earliest transaction date: {earliest_date.date()}")
print(f"Latest transaction date: {latest_date.date()}\n")

# 2 & 3. Calculate the days between each transaction and the first transaction, 
# and assign it to a new column
sales["Elapsed Days"] = (sales["Date"] - earliest_date).dt.days

print("Data with Elapsed Days:\n", sales[["Date", "Product", "Revenue", "Elapsed Days"]], "\n")

````

````python
# 4. Identify the oldest and most recent sales records (returning the entire row)
# idxmin() and idxmax() find the index of the minimum and maximum dates
oldest_record = sales.loc[sales["Date"].idxmin()]
most_recent_record = sales.loc[sales["Date"].idxmax()]

print("--- Oldest Sales Record ---")
print(oldest_record, "\n")

print("--- Most Recent Sales Record ---")
print(most_recent_record)
````

### 📝 Observation

- Datetime objects support arithmetic operations such as subtraction and duration calculations.
- Time differences help measure business intervals and identify trends over time.
- Date arithmetic is widely used for customer retention, delivery tracking, and performance analysis.

# 🧩 Mini Challenge – Employee Attendance Analysis

## Scenario

An HR department maintains employee attendance records.

Create a dataset containing:

- Employee
- Department
- Check-In Date
- Hours Worked

Using Pandas:

- Convert Check-In Date into datetime.
- Extract Month and Day Name.
- Find employees who checked in after a specified date.
- Calculate the average working hours for each month.
- Count attendance records for each weekday.

Finally,

Write a short summary explaining how datetime features improved attendance analysis.

````python
employee_data = pd.DataFrame({
    "Employee": [
        "Alice", 
        "Bob", 
        "Charlie", 
        "Diana", 
        "Evan", 
        "Fiona", 
        "George", 
        "Hannah", 
        "Ian", 
        "Julia"
    ],
    "Department": [
        "IT", 
        "HR", 
        "Sales", 
        "IT", 
        "Marketing", 
        "Sales", 
        "IT", 
        "HR", 
        "Marketing", 
        "Sales"
    ],
    "Check-In Date": [
        "2026-07-15",
        "2026-07-15",
        "2026-07-18",
        "2026-07-22",
        "2026-08-05",
        "2026-08-09",
        "2026-08-14",
        "2026-08-20",
        "2026-08-25",
        "2026-09-02"
    ],
    "Hours Worked": [
        8.0, 
        7.5, 
        9.0, 
        8.5, 
        4.0, 
        8.0, 
        10.0, 
        7.0, 
        6.5, 
        8.0
    ]
})
print(employee_data)
````

````python
#Convert Check-In Date into datetime.
employee_data["Check-In Date"] = pd.to_datetime(employee_data["Check-In Date"])
#Extract Month and Day Name.
employee_data["Month"] = employee_data["Check-In Date"].dt.month_name()
employee_data["Day"] = employee_data["Check-In Date"].dt.day_name()
print(employee_data)
````

````python
#Find employees who checked in after a specified date.
print("employees who checked in after 2026-08-09 : ",employee_data[employee_data["Check-In Date"]>"2026-08-09"].Employee.to_list())
#Calculate the average working hours for each month.
print("\naverage working hours for each month : \n",employee_data.groupby("Month").agg({"Hours Worked" : "mean"}))
#Count attendance records for each weekday.
print("\nattendance records for each weekday : \n",employee_data.groupby("Day").agg({"Employee" : "count"}))
````

# 🌟 Analyst Extension Challenge – Answers

### 1. Convert a text column into actual dates.

**Answer:** `pd.to_datetime()`

**Reason:** Converts string or object data into Pandas datetime objects for time-based analysis.

---

### 2. Extract the month from each transaction.

**Answer:** `.dt.month`

**Reason:** Retrieves the numeric month from each datetime value.

---

### 3. Find all transactions after March 1.

**Answer:** Boolean filtering using datetime comparison

Example:

```python
df[df["Date"] > "2026-03-01"]
```

**Reason:** Datetime columns support chronological comparison operators.

---

### 4. Calculate the number of days between two dates.

**Answer:** Date subtraction (`Timedelta`)

Example:

```python
df["Date"] - df["Date"].min()
```

**Reason:** Subtracting datetime values produces a Timedelta object representing the duration.

---

### 5. Display the weekday for every transaction.

**Answer:** `.dt.day_name()`

**Reason:** Returns the weekday corresponding to each transaction date.

---

### 6. Identify the quarter for each transaction.

**Answer:** `.dt.quarter`

**Reason:** Determines the quarter (1–4) for every datetime value.

---

### 7. Calculate monthly revenue.

**Answer:** `groupby(df["Date"].dt.month)["Revenue"].sum()`

**Reason:** Groups transactions by month before calculating total revenue.

---

### 8. Find the earliest transaction date.

**Answer:** `min()`

Example:

```python
df["Date"].min()
```

**Reason:** Returns the earliest datetime present in the dataset.
