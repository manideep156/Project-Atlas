# 📖 Part 2 — Filtering DataFrames

### Objective
Retrieve records based on specific business conditions.

Filtering is one of the most frequently used operations in data processing.

For example, an HR department might want to retrieve only employees whose salaries exceed ₹60,000.

PySpark provides two commonly used methods:

- `filter()`
- `where()`

Both perform the same filtering operation.

### 2.1 Basic Filtering

Syntax:

    dataframe.filter(condition)

Example: Retrieve employees earning more than ₹60,000.



```python


employee_df.filter(
        col("salary") > 60000
    ).show()
```


### 2.2 Using where()

The `where()` method is an alternative to `filter()`.

    employee_df.where(
        col("department") == "IT"
    ).show()

### 2.3 Comparison Operators

PySpark supports common comparison operators:

| Operator | Meaning |
|---|---|
| `==` | Equal to |
| `!=` | Not equal to |
| `>` | Greater than |
| `<` | Less than |
| `>=` | Greater than or equal to |
| `<=` | Less than or equal to |



```python


employee_df.filter(col("salary") >= 50000).show()
employee_df.filter(col("experience") < 5).show()
employee_df.filter(col("department") != "HR").show()
```


### 2.4 Combining Multiple Conditions

PySpark supports logical conditions.

| Operator | Meaning |
|---|---|
| `&` | AND |
| `|` | OR |
| `~` | NOT |

Important: Always place individual conditions inside parentheses.



```python


    # Employees in IT earning more than 50000
employee_df.filter(
        (col("department") == "IT") &
        (col("salary") > 50000)
    ).show()

    # Employees from Chennai or Bangalore
employee_df.filter(
        (col("city") == "Chennai") |
        (col("city") == "Bangalore")
    ).show()

    # Exclude employees belonging to HR
employee_df.filter(
        ~(col("department") == "HR")
    ).show()
```


### 2.5 Using isin()

`isin()` checks whether a column value belongs to a given collection.


```python

employee_df.filter(
        col("department").isin("IT", "Finance")
    ).show()
```


### 2.6 Using between()

`between()` checks whether a value falls within an inclusive range.

```python


employee_df.filter(
        col("salary").between(45000, 65000)
    ).show()
```


### 🔍 Observation

- `filter()` and `where()` are interchangeable.
- Multiple conditions can be combined using `&` and `|`.
- `isin()` is useful when checking against multiple possible values.
- `between()` includes both boundary values.


# 📖 Part 3 — Selecting Columns & Creating Expressions

### Objective
Select relevant columns and create calculated values from existing columns.

In enterprise data pipelines, we rarely need every column from a source dataset.

For example, a finance report might require only employee name, department, and annual salary.

### 3.1 Using select()

`select()` retrieves specific columns.



```python


employee_df.select(
        "name",
        "department",
        "city"
    ).show()
```


### 3.2 Creating Calculated Columns

PySpark allows arithmetic operations directly on columns.

Suppose we want to calculate annual salary.

    annual_salary = monthly_salary * 12

```python

employee_df.select(
        "name",
        "salary",
        (col("salary") * 12).alias("annual_salary")
    ).show()
```


### 3.3 Using alias()

`alias()` assigns a temporary name to a selected expression.


```python


employee_df.select(
        col("name").alias("employee_name"),
        col("salary").alias("monthly_salary"),
        (col("salary") * 12).alias("annual_salary")
    ).show()
```


### 3.4 Using expr()

`expr()` allows us to write SQL-style expressions inside DataFrame transformations.



```python
from pyspark.sql.functions import expr

employee_df.select(
        "name",
        expr("salary * 12 AS annual_salary"),
        expr("experience + 1 AS projected_experience")
    ).show()
```


### 🔍 Observation

- `select()` controls which columns appear in the result.
- `alias()` renames an expression in the output.
- `expr()` allows SQL-style expressions.
- These operations do not modify the original DataFrame.


# 📖 Part 4 — Removing Duplicate Records

### Objective
Understand how duplicate records are identified and removed.

Duplicates are common in enterprise datasets because of repeated ingestion, source-system issues, or multiple data loads.

PySpark provides:

- `distinct()`
- `dropDuplicates()`

### 4.1 Using distinct()

`distinct()` removes duplicate rows by considering all columns.


```python

employee_df.select(
        "department"
    ).distinct().show()
```


### 4.2 Using dropDuplicates()

`dropDuplicates()` allows us to specify the columns used to identify duplicates.

```python
employee_df.dropDuplicates(
        ["department"]
    ).show()
```


The above operation retains one record for each department.

Note: When duplicate rows contain different values in other columns, do not assume which row will be retained. If a specific record must be selected, define a deterministic rule using sorting and a window function.



### 4.3 Creating a Dataset with Duplicates


```python

duplicate_data = [
        (101, "Arjun", "IT", 55000),
        (102, "Priya", "HR", 42000),
        (103, "Rahul", "IT", 65000),
        (101, "Arjun", "IT", 55000),
        (104, "Sneha", "Finance", 48000),
        (102, "Priya", "HR", 42000)
    ]

duplicate_columns = [
        "employee_id",
        "name",
        "department",
        "salary"
    ]

duplicate_df = spark.createDataFrame(
        duplicate_data,
        duplicate_columns
    )

duplicate_df.show()
```


### 4.4 Removing Duplicates



```python
    # Remove fully duplicated rows
duplicate_df.distinct().show()

    # Remove duplicates using employee_id
duplicate_df.dropDuplicates(
        ["employee_id"]
    ).show()
```


### 🔍 Observation

- `distinct()` considers the entire row.
- `dropDuplicates()` can consider selected columns.
- Deduplication based on a business key is common in data engineering.
- If conflicting records share the same key, an explicit selection rule is needed.


# 📖 Part 6 — Grouping & Aggregations

### Objective
Summarize records using grouping and aggregate functions.

This is one of the most important concepts in data analytics and data engineering.

Imagine an organization with thousands of employees.

Management may not want individual employee records. Instead, it may want:
- Total employees per department.
- Average salary per department.
- Highest salary per department.
- Total salary expenditure per city.

This is where `groupBy()` and aggregation functions become useful.

### 6.1 Understanding groupBy()

`groupBy()` groups records based on one or more columns.

Syntax:

    dataframe.groupBy("column").agg(...)

### 6.2 Counting Records


```python

from pyspark.sql.functions import count

employee_df.groupBy(
        "department"
    ).agg(
        count("*").alias("employee_count")
    ).show()
```


### 6.3 Average Salary


```python

from pyspark.sql.functions import avg

employee_df.groupBy(
        "department"
    ).agg(
        avg("salary").alias("average_salary")
    ).show()
```


### 6.4 Sum of Salary


```python
from pyspark.sql.functions import sum
employee_df.groupBy(
        "department"
    ).agg(
        sum("salary").alias("total_salary")
    ).show()
```


### 6.5 Minimum & Maximum


```python
from pyspark.sql.functions import min, max
employee_df.groupBy(
        "department"
    ).agg(
        min("salary").alias("minimum_salary"),
        max("salary").alias("maximum_salary")
    ).show()
```


### 6.6 Multiple Aggregations

We can calculate multiple metrics in a single aggregation.

```python
employee_df.groupBy(
        "department"
    ).agg(
        count("*").alias("employee_count"),
        avg("salary").alias("average_salary"),
        sum("salary").alias("total_salary"),
        min("salary").alias("minimum_salary"),
        max("salary").alias("maximum_salary")
    ).show()
```


### 6.7 Grouping by Multiple Columns

We can group records using more than one column.

For example, calculate the average salary for each department in each city.

```python
employee_df.groupBy(
        "department",
        "city"
    ).agg(
        avg("salary").alias("average_salary"),
        count("*").alias("employee_count")
    ).show()
```


### 6.8 Overall Aggregations

We can aggregate the entire DataFrame without grouping.


```python

employee_df.agg(
        count("*").alias("total_employees"),
        avg("salary").alias("overall_average_salary"),
        max("salary").alias("highest_salary"),
        min("salary").alias("lowest_salary")
    ).show()
```


### 🔍 Observation

- `groupBy()` creates logical groups.
- `agg()` applies one or more aggregate functions.
- `count()` counts records.
- `sum()` calculates totals.
- `avg()` calculates averages.
- `min()` and `max()` identify extreme values.
- Grouping by multiple columns produces more granular summaries.


# 📖 Part 7 — Combining Transformations into Pipelines

### Objective
Combine filtering, selecting, expressions, sorting, and aggregations into a single analytical workflow.

In enterprise environments, transformations are usually chained together rather than executed independently.

Example business requirement:

Retrieve IT employees earning more than ₹50,000, calculate their annual salary, and sort the result by annual salary.



```python
from pyspark.sql.functions import col

it_salary_analysis = (
        employee_df
        .filter(
            (col("department") == "IT") &
            (col("salary") > 50000)
        )
        .select(
            "name",
            "department",
            "salary",
            (col("salary") * 12).alias("annual_salary")
        )
        .orderBy(
            col("annual_salary").desc()
        )
    )

it_salary_analysis.show()
```


### 7.1 Department-Level Analytical Pipeline



```python
department_summary = (
        employee_df
        .groupBy("department")
        .agg(
            count("*").alias("employee_count"),
            avg("salary").alias("average_salary"),
            sum("salary").alias("total_salary")
        )
        .orderBy(
            col("total_salary").desc()
        )
    )

department_summary.show()

```

### 🔍 Observation

A DataFrame transformation pipeline makes the workflow easier to read and maintain.

Remember that Spark transformations are generally lazy. Spark builds an execution plan and performs the work when an action such as `show()`, `count()`, or `collect()` is invoked.

# 📖 Part 11 — Common Mistakes & Things to Remember

### 1. Incorrect logical operators

Incorrect:

    employee_df.filter(
        col("salary") > 50000 and col("experience") > 3
    )

Correct:

    employee_df.filter(
        (col("salary") > 50000) &
        (col("experience") > 3)
    )

Use `&`, `|`, and `~` for PySpark column expressions.

### 2. Forgetting parentheses

Always wrap individual conditions when combining them.

### 3. Confusing distinct() and dropDuplicates()

- `distinct()` considers every column.
- `dropDuplicates()` can consider selected columns.

### 4. Forgetting to aggregate after groupBy()

Incorrect:

    employee_df.groupBy("department").show()

Correct:

    employee_df.groupBy("department").count().show()

### 5. Assuming transformations execute immediately

Spark uses lazy evaluation for transformations.

Actions trigger execution.

### 6. Ignoring sorting costs

Global sorting can involve data shuffling and may be expensive on large datasets.

Use sorting when the business requirement actually needs ordered results.

### 7. Using collect() unnecessarily

`collect()` brings all records to the driver.

Avoid using it on large DataFrames unless the result is known to be small.
