# 📖 Part 1 – Apache Spark and PySpark

**Objective:** Understand what Spark and PySpark are at a beginner level.

Apache Spark is a data-processing engine designed to process large datasets efficiently. It can divide work across multiple computing resources instead of depending entirely on one processor or one machine.

PySpark is the Python interface for Apache Spark. It allows Spark operations to be written using Python while the Spark engine performs the underlying processing.

For today's local setup, Spark will run only on my computer. Later, similar PySpark code can run on clusters and platforms such as Azure Databricks.
