# 📖 Part 1 – Understanding Spark Architecture

**Objective:** Understand how Spark processes data.

Apache Spark uses a distributed processing architecture to divide computational work into smaller tasks.

Its primary components include:

- **Driver:** Coordinates the Spark application and manages execution.
- **Executors:** Execute tasks and process data.
- **Cluster Manager:** Allocates computing resources to the Spark application.
- **Partitions:** Divide datasets into smaller logical sections that can be processed in parallel.

For our current learning environment, Spark will run locally using the available computer resources.

Later, we will explore how the same concepts apply to distributed environments such as Azure Databricks.



# 📖 Part 2 – `SparkSession`

**Objective:** Understand how a PySpark application starts.

A `SparkSession` is the entry point for interacting with Apache Spark through PySpark.

It allows us to:

- Create Spark DataFrames.
- Read datasets.
- Perform DataFrame operations.
- Execute Spark SQL queries.

Let's create our first SparkSession for today's learning session.


```python
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .appName("ProjectAtlas_Day22")
    .master("local[*]")
    .getOrCreate()
)

print(spark.version)
```


# 📖 Part 3 – Introduction to Spark DataFrames

**Objective:** Understand how data is represented and accessed in PySpark.

A Spark DataFrame is a distributed collection of data organized into named columns.

It is conceptually similar to a Pandas DataFrame, but Spark DataFrames are designed to support distributed data processing.

Let's create a small employee dataset.


```python


data = [
    (1, "Alice", "IT", 45000),
    (2, "Bob", "HR", 35000),
    (3, "Charlie", "IT", 55000),
    (4, "David", "Finance", 40000),
    (5, "Eva", "HR", 38000)
]

columns = ["id", "name", "department", "salary"]

df = spark.createDataFrame(data, columns)

df.show()

```


# 📖 Part 4 – Exploring Spark DataFrames

**Objective:** Learn basic commands for inspecting a Spark DataFrame.

| Method | Purpose |
|---|---|
| `show()` | Displays DataFrame records |
| `printSchema()` | Displays the DataFrame schema |
| `columns` | Returns column names |
| `dtypes` | Returns column names and data types |
| `count()` | Returns the number of records |

Try executing these commands individually and observe their outputs.


```python


df.show()

df.printSchema()

df.columns

df.dtypes

df.count()
```



# 📖 Part 5 – Basic DataFrame Operations

**Objective:** Perform simple operations on Spark DataFrames.



```python

# Selecting specific columns
df.select("name", "salary").show()

# Filtering records
df.filter(df.salary > 40000).show()

# Sorting records
df.orderBy("salary", ascending=False).show()
```
